##  Question 1

```python
def calculator():
    print("Select Operation:")
    print("1. Addition       (+)")
    print("2. Subtraction    (-)")
    print("3. Multiplication (*)")
    print("4. Division       (/)")
    print("5. Power          (**)")

    choice = input("Enter choice (1/2/3/4/5): ")

    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    if choice == '1':
        result = num1 + num2
        print(f"Result: {num1} + {num2} = {result}")

    elif choice == '2':
        result = num1 - num2
        print(f"Result: {num1} - {num2} = {result}")

    elif choice == '3':
        result = num1 * num2
        print(f"Result: {num1} * {num2} = {result}")

    elif choice == '4':
        if num2 == 0:
            print(" Error! Division by zero is not allowed.")
        else:
            result = num1 / num2
            print(f"Result: {num1} / {num2} = {result}")

    elif choice == '5':
        result = num1 ** num2
        print(f"Result: {num1} ** {num2} = {result}")

    else:
        print(" Invalid choice! Please select a valid operation.")

calculator()
```

##  Question 2: 

```python
# Mixed list containing different data types
my_list = ["hello", 42, "world", 3.14, "python", True, "jawan"]

print("List:", my_list)
print("-" * 40)

found_numeric = False

for item in my_list:
    if isinstance(item, (int, float)) and not isinstance(item, bool):
        print(f" Numeric value found: {item}  (type: {type(item).__name__})")
        found_numeric = True

print("-" * 40)
if found_numeric:
    print(" The list contains numeric value(s).")
else:
    print(" No numeric values found in the list.")
```

## Question 4: 

```python
data = {
    "name": "Ahmed",
    "marks": 85,
    "city": "Lahore",
    "fee": 1500.50,
    "age": 22,
    "grade": "A",
    "bonus": 200
}

print("Dictionary:")
print(data)
print("-" * 40)

total = 0

print("Numeric values found:")
for key, value in data.items():
    if isinstance(value, (int, float)) and not isinstance(value, bool):
        print(f"  {key}: {value}")
        total += value

print("-" * 40)
print(f"Sum of all numeric values = {total}")
```

##  Question 5

```python
# List with duplicate values
my_list = [10, 20, 30, 20, 40, 10, 50, 30, 60, 10]

print("Original List:")
print(my_list)
print("-" * 40)

duplicates = []
seen = []

for item in my_list:
    if item in seen:
        if item not in duplicates:
            duplicates.append(item)
    else:
        seen.append(item)

print("Duplicate Values Found:")
if duplicates:
    print(f"   Duplicates: {duplicates}")
else:
    print("  No duplicate values found.")

print("-" * 40)

# Count how many times each duplicate appears
print("Duplicate Count:")
for item in duplicates:
    count = my_list.count(item)
    print(f"  Value '{item}' appears {count} times")
```

```python
# Sample dictionary
employee = {
    "name": "Sara",
    "department": "IT",
    "salary": 50000,
    "experience": 3
}

print("Dictionary:")
print(employee)
print("-" * 40)

# Keys to check
keys_to_check = ["name", "age", "salary", "address"]

for key in keys_to_check:
    if key in employee:
        print(f" Key '{key}' EXISTS in the dictionary. Value = {employee[key]}")
    else:
        print(f" Key '{key}' does NOT exist in the dictionary.")

print("-" * 40)

# Interactive check
user_key = input("\nEnter a key to check: ")
if user_key in employee:
    print(f" Key '{user_key}' found! Value = {employee[user_key]}")
else:
    print(f" Key '{user_key}' does not exist in the dictionary.")
```
