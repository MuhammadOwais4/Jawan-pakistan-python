## Q7. Write a program which takes 5 inputs from user for different subject's marks, total it and generate mark sheet using grades

```python
math_marks     = float(input("Enter marks for Math (out of 100): "))
english_marks  = float(input("Enter marks for English (out of 100): "))
science_marks  = float(input("Enter marks for Science (out of 100): "))
urdu_marks     = float(input("Enter marks for Urdu (out of 100): "))
computer_marks = float(input("Enter marks for Computer (out of 100): "))

total      = math_marks + english_marks + science_marks + urdu_marks + computer_marks
percentage = total / 5

if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

print("")
print("========================================")
print("             MARK SHEET                 ")
print("========================================")
print("Math      :", math_marks)
print("English   :", english_marks)
print("Science   :", science_marks)
print("Urdu      :", urdu_marks)
print("Computer  :", computer_marks)
print("----------------------------------------")
print("Total     :", total, "/ 500")
print("Percentage:", percentage, "%")
print("Grade     :", grade)
print("========================================")
```

## Q8. Write a Python program which take input from user and identify that the given number is even or odd

```python
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(number, "is an EVEN number.")
else:
    print(number, "is an ODD number.")
```
