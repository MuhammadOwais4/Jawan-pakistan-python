## Q7. Take 5 subject marks, total them, and generate a mark sheet with grades

```python
subjects = ["Math", "English", "Science", "Urdu", "Computer"]
marks = []

# Taking input for each subject
for subject in subjects:
    mark = float(input(f"Enter marks for {subject} (out of 100): "))
    marks.append(mark)

# Grade function
def get_grade(m):
    if m >= 90:
        return "A+"
    elif m >= 80:
        return "A"
    elif m >= 70:
        return "B"
    elif m >= 60:
        return "C"
    elif m >= 50:
        return "D"
    else:
        return "F"

total = sum(marks)
percentage = total / len(subjects)
overall_grade = get_grade(percentage)

# Print Mark Sheet
print("\n" + "="*40)
print("         MARK SHEET")
print("="*40)
print(f"{'Subject':<15} {'Marks':>6} {'Grade':>6}")
print("-"*40)
for i in range(len(subjects)):
    print(f"{subjects[i]:<15} {marks[i]:>6.1f} {get_grade(marks[i]):>6}")
print("-"*40)
print(f"{'Total':<15} {total:>6.1f}")
print(f"{'Percentage':<15} {percentage:>5.1f}%")
print(f"{'Overall Grade':<15} {overall_grade:>6}")
print("="*40)
```

## Q8. Take input from user and identify if the number is Even or Odd

```python
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(f"{number} is an EVEN number.")
else:
    print(f"{number} is an ODD number.")
```
